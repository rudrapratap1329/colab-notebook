```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url="https://raw.githubusercontent.com/plotly/datasets/master/gapminderDataFiveYear.csv"

df=pd.read_csv(url)


print(df.head())
print(df.shape)
print(df.columns)
print(df.info())
```

```python
print("Missing Values")
print(df.isna().sum())
```

```python
print("year range")
```

```python
import matplotlib.pyplot as plt

# 1. Calculate the average life expectancy for each continent
avg_life_exp = df.groupby('continent')['lifeExp'].mean().sort_values()

# 2. Initialize the plot style and size
plt.figure(figsize=(8, 5))

# 3. Create the bar chart
plt.bar(avg_life_exp.index, avg_life_exp.values, color='skyblue', edgecolor='black')

# 4. Add titles and axis labels
plt.title('Average Life Expectancy by Continent', fontsize=14, fontweight='bold')
plt.xlabel('Continent', fontsize=12)
plt.ylabel('Average Life Expectancy (Years)', fontsize=12)

# 5. Add a grid behind the bars for easier reading
plt.grid(axis='y', linestyle='--', alpha=0.7)

# 6. Display the chart
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Filter the dataset for the three specific countries
target_countries = ['India', 'China', 'United States']
filtered_df = df[df['country'].isin(target_countries)]

# 2. Initialize the plot size
plt.figure(figsize=(10, 6))

# 3. Loop through each country and plot its line over time
for country in target_countries:
    country_data = filtered_df[filtered_df['country'] == country]
    plt.plot(country_data['year'], country_data['lifeExp'], marker='o', label=country, linewidth=2)

# 4. Add titles, labels, and visual enhancements
plt.title('Life Expectancy Over Time (1952–2007)', fontsize=14, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Life Expectancy (Years)', fontsize=12)

# 5. Add a legend and grid for readability
plt.legend(title="Country", fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)

# 6. Display the chart
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Filter the dataset for the three target countries
target_countries = ['India', 'China', 'United States']
filtered_df = df[df['country'].isin(target_countries)]

# 2. Set up distinct colors for each country
colors = {'India': '#FF9933', 'China': '#DE2910', 'United States': '#002868'}

# 3. Initialize the plot size
plt.figure(figsize=(10, 6))

# 4. Loop through and plot each country as a separate group
for country in target_countries:
    country_data = filtered_df[filtered_df['country'] == country]
    plt.scatter(
        country_data['gdpPercap'],
        country_data['lifeExp'],
        color=colors[country],
        label=country,
        s=100,            # Marker size
        alpha=0.8,        # Transparency
        edgecolors='black'# Border around points
    )

# 5. Add titles, labels, and formatting
plt.title('GDP Per Capita vs. Life Expectancy', fontsize=14, fontweight='bold')
plt.xlabel('GDP Per Capita (USD)', fontsize=12)
plt.ylabel('Life Expectancy (Years)', fontsize=12)

# 6. Add legend and grid
plt.legend(title="Country", fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)

# 7. Display the chart
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Initialize the plot size
plt.figure(figsize=(10, 6))

# 2. Group by continent to color-code the global data cleanly
for continent, group in df.groupby('continent'):
    plt.scatter(
        group['gdpPercap'],
        group['lifeExp'],
        label=continent,
        alpha=0.6,        # Translucent dots help see overlapping data
        edgecolors='none',
        s=40             # Base marker size
    )

# 3. Apply a Logarithmic Scale to the X-axis (Crucial for global GDP distribution)
plt.xscale('log')

# 4. Add titles and clear axis labels
plt.title('Global GDP Per Capita vs. Life Expectancy', fontsize=14, fontweight='bold')
plt.xlabel('GDP Per Capita (USD) - Log Scale', fontsize=12)
plt.ylabel('Life Expectancy (Years)', fontsize=12)

# 5. Format the X-axis ticks to display normal numbers instead of scientific notation
from matplotlib.ticker import ScalarFormatter
plt.gca().xaxis.set_major_formatter(ScalarFormatter())

# 6. Add legend and structural grid
plt.legend(title="Continent", fontsize=10, loc='lower right')
plt.grid(True, which="both", linestyle='--', alpha=0.4)

# 7. Display the chart
plt.tight_layout()
plt.show()
```
